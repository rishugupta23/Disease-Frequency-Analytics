Outliers detection using IQR method

```python
# finding outliers in BMI
q1 = df['BMI'].quantile(0.25)
q3 = df['BMI'].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5*iqr
upper = q3 + 1.5*iqr

outliers = df[(df['BMI'] < lower) | (df['BMI'] > upper)]
print(outliers)
```

```python
# finding outliers in Blood_Glucose_mg_dL
q1 = df['Blood_Glucose_mg_dL'].quantile(0.25)
q3 = df['Blood_Glucose_mg_dL'].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5*iqr
upper = q3 + 1.5*iqr

outliers = df[(df['Blood_Glucose_mg_dL'] < lower) | (df['Blood_Glucose_mg_dL'] > upper)]
print(outliers)
```

```python
# finding outliers in HbA1c
q1 = df['HbA1c_%'].quantile(0.25)
q3 = df['HbA1c_%'].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5*iqr
upper = q3 + 1.5*iqr

outliers = df[(df['HbA1c_%'] < lower) | (df['HbA1c_%'] > upper)]
print(outliers)
```

```python
# finding outliers iTotal_Cholesterol_mg_dL
q1 = df['Total_Cholesterol_mg_dL'].quantile(0.25)
q3 = df['Total_Cholesterol_mg_dL'].quantile(0.75)
iqr = q3 - q1
lower = q1 - 1.5*iqr
upper = q3 + 1.5*iqr

outliers = df[(df['Total_Cholesterol_mg_dL'] < lower) | (df['Total_Cholesterol_mg_dL'] > upper)]
print(outliers)
```

```python
df["Primary_Diagnosis"] = df["Primary_Diagnosis"].replace(
    r"^\s*$", np.nan, regex=True
)
df["Primary_Diagnosis"] = df["Primary_Diagnosis"].fillna(
    "Unknown / Not recorded"
)
print("Missing values after handling:")
print(df["Primary_Diagnosis"].isnull().sum())

print("\nDiagnosis-wise record counts:")
print(df["Primary_Diagnosis"].value_counts())
```

```python
df["Blood_Glucose_mg_dL"] = df["Blood_Glucose_mg_dL"].replace(
    999, np.nan
)
glucose_median = df["Blood_Glucose_mg_dL"].median()
df["Blood_Glucose_mg_dL"] = df["Blood_Glucose_mg_dL"].fillna(glucose_median)
print("Missing glucose values:", df["Blood_Glucose_mg_dL"].isnull().sum())
print("Median used:", glucose_median)
```

3) Regional Disease Distribution of Type 2 Diabetes Patients in India

```python
type_2_diabetes = df[df['Primary_Diagnosis']=='Type 2 Diabetes'].groupby('Region')['Patient_ID'].count().reset_index(name='No of Patient')
print(type_2_diabetes)
```

```python
import folium
```

```python
m = folium.Map(location=[22.5, 80],zoom_start=5)
m
```

```python
import geopandas as gpd
```

```python
india_map = gpd.read_file(
    '/content/Admin2.shp',
    on_invalid='fix'
)

india_map.head()
```

```python
print(india_map.columns)
```

```python
print(india_map['ST_NM'].unique())
```

```python
diabetes_map = india_map[
    india_map['ST_NM'].isin(type_2_diabetes['Region'])
].merge(
    type_2_diabetes,
    left_on='ST_NM',
    right_on='Region',
    how='left'
)

diabetes_map[['ST_NM', 'No of Patient']]
```

```python
m_diabetes_clean = folium.Map(
    location=[22.5, 80],
    zoom_start=5,
    tiles=None
)
```

```python
folium.Choropleth(
    geo_data=diabetes_map.__geo_interface__,
    data=diabetes_map,
    columns=['ST_NM', 'No of Patient'],
    key_on='feature.properties.ST_NM',
    fill_color='YlOrRd',
    legend_name='Recorded Type 2 Diabetes Patients'
).add_to(m_diabetes_clean)

m_diabetes_clean
```

```python
m_diabetes_clean.save('/content/type2_diabetes_choropleth.html')
```

The choropleth map shows regional variation in the recorded Type 2 Diabetes patients within the dataset. Darker shades indicate relatively higher concentrations of recorded patients, while lighter shades indicate relatively lower concentrations.

6. Disease vs Treatment Outcome

```python
outcomes = (df.groupby(['Primary_Diagnosis','Treatment_Outcome'])['Patient_ID'].count().reset_index(name='Patient_Count'))
print(outcomes)
```

```python
sns.set_theme(style='whitegrid')

outcomes_pivot = outcomes.pivot(
    index='Primary_Diagnosis',
    columns='Treatment_Outcome',
    values='Patient_Count'
).fillna(0)

outcomes_pivot.plot(
    kind='bar',
    stacked=True,
    figsize=(12, 6)
)

plt.title('Disease vs Treatment Outcome')
plt.xlabel('Primary Diagnosis')
plt.ylabel('Number of Patient Records')
plt.xticks(rotation=90, ha='right')
plt.legend(title='Treatment Outcome')
plt.tight_layout()
plt.show()
```

10. Monthly Disease Frequency Analysis

```python
df['Visit_Date'] = pd.to_datetime(df['Visit_Date'],errors='coerce')
print(df['Visit_Date'].dtype)
```

```python
print("Earliest Visit Date:", df['Visit_Date'].min())
print("Latest Visit Date:", df['Visit_Date'].max())
print("Missing Visit Dates:", df['Visit_Date'].isnull().sum())
```

```python
df['Visit_Month'] = df['Visit_Date'].dt.to_period('M').astype(str)
print(df[['Visit_Date', 'Visit_Month']].head(10))
```

```python
monthly_records = (
    df.dropna(subset=['Visit_Date'])
    .groupby('Visit_Month')['Patient_ID']
    .count()
    .reset_index(name='Patient_Count')
)

print(monthly_records)
```
